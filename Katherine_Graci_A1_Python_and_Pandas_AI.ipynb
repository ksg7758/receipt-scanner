### Q5 — Why logs show up everywhere in finance

There is a trick worth knowing: for numbers close to 1, the percent change
$(x-y)/y$ is well approximated by the difference in logs, $\log x - \log y$.

Verify it with the numbers below — compute both and compare. Then try it again
with `x = 2.0, y = 1.0` and report how well the approximation holds.

One sentence: when is it safe to use, and when is it not?

````python
import math

x, y = 1.05, 1.02

# Calculate percent change
percent_change_1 = (x - y) / y

# Calculate difference in logs
diff_logs_1 = math.log(x) - math.log(y)

print(f"For x={x}, y={y}:")
print(f"  Percent Change: {percent_change_1:.6f}")
print(f"  Difference in Logs: {diff_logs_1:.6f}")

x_large, y_large = 2.0, 1.0

# Calculate percent change for larger difference
percent_change_2 = (x_large - y_large) / y_large

# Calculate difference in logs for larger difference
diff_logs_2 = math.log(x_large) - math.log(y_large)

print(f"\nFor x={x_large}, y={y_large}:")
print(f"  Percent Change: {percent_change_2:.6f}")
print(f"  Difference in Logs: {diff_logs_2:.6f}")
````

When `x` and `y` are close to 1 (e.g., 1.05 and 1.02), the percent change and the difference in logs are very similar, indicating a good approximation. However, when `x` and `y` are far apart (e.g., 2.0 and 1.0), the approximation breaks down, and the values are no longer close.

### Q7 — Missing values that are not missing

The header text says: *"All missing values are indicated by -99.99 or -999."*

Find them and replace them with proper `NaN`. Then report what fraction of the
table was missing.

> **⚠️** If you skip this, `-99.99` is a perfectly valid number and every mean,
> standard deviation and correlation you compute will quietly include it as a
> **−99.99% monthly return**. Nothing will error.

````python
import numpy as np

# Replace -99.99 and -999 with NaN
ind_cleaned = ind.replace([-99.99, -999], np.nan)

# Calculate total number of cells
total_cells = ind_cleaned.size

# Calculate number of missing values
missing_values = ind_cleaned.isna().sum().sum()

# Calculate fraction of missing values
fraction_missing = missing_values / total_cells

print(f"Total cells in the table: {total_cells}")
print(f"Number of missing values: {missing_values}")
print(f"Fraction of the table that was missing: {fraction_missing:.4f}")
````

### Q9 — Excess returns

Load the `Market_proxy` sheet the same way — it has `Mkt-RF` and `RF`. Watch the
units here too.

Then build excess returns two ways:

1. For **one** industry, `Agric`, subtract `RF` month by month. Print its mean.
2. For **all 49 at once**, in a single line, producing a DataFrame called `inde`.

> **🤖 Worth asking the AI:** *"subtract a Series from every column of a
> DataFrame, aligning on the index"* — and then check that the row count did not
> change. Getting the axis wrong here produces a table full of NaN, or worse, a
> table that looks fine and is transposed.

````python
# Load the 'Market_proxy' sheet, explicitly setting header to the 4th row (index 3)
market = pd.read_excel(URL, sheet_name='Market_proxy', header=3)

# The current DataFrame 'market' has an unexpected name for the first column (the date column).
# It is named 'The 1-month TBill return is from Ibbotson and Associates Inc.'
# We need to rename this column to 'Date'.
market = market.rename(columns={market.columns[0]: 'Date'})

# Clean market data similar to industries
# Convert the 'Date' column to datetime and set it as the index
market.index = pd.to_datetime(market['Date'], format='%Y%m') + pd.offsets.MonthEnd(0)
market = market.drop(columns=['Date'])

# Replace missing values (if any) and convert to decimal
# The Mkt-RF and RF columns might also contain -99.99 or -999, so apply replacement to all relevant columns
market.iloc[:, :] = market.iloc[:, :].replace([-99.99, -999], np.nan)
market.iloc[:, :] = market.iloc[:, :] / 100


# Ensure both DataFrames have the same date range and columns aligned
# We will align ind_cleaned and market on their common index

# Extract RF (Risk-Free rate)
RF = market['RF']

# 1. For one industry, Agric, subtract RF month by month
ex_ret_agric = ind_cleaned['Agric'] - RF
print(f"Mean excess return for Agric: {ex_ret_agric.mean():.5f}")

# 2. For all 49 at once, in a single line, producing a DataFrame called inde
# We need to ensure that the RF Series is broadcast correctly across the DataFrame.
# Pandas automatically aligns by index, so simple subtraction works.
inde = ind_cleaned.subtract(RF, axis=0)

print(f"\nShape of the excess returns DataFrame (inde): {inde.shape}")
print("First 3 rows of excess returns for a few industries:")
print(inde.head(3))
````
